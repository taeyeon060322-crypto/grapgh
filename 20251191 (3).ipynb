```python
import networkx as nx
import matplotlib.pyplot as plt
import platform

system_name = platform.system()
if system_name == 'Windows':
    font_name = 'Malgun Gothic'
elif system_name == 'Darwin':
    font_name = 'AppleGothic'
else:
    font_name = 'NanumGothic'


plt.rc('font', family=font_name)
plt.rcParams['axes.unicode_minus'] = False

# 2. 노드 및 음식점 라벨 설정
labels = {
    1: "멜트", 2: "방목", 3: "용산토리", 4: "동녘",
    5: "초이노미", 6: "이화만두", 7: "상상홍익점", 8: "팔도실비집"
}

# 인접 행렬 데이터 (이동 시간 가중치)
matrix = {
    1: {2:34, 3:15, 4:21, 5:27, 6:18, 7:19, 8:23},
    2: {1:34, 3:38, 4:30, 5:37, 6:23, 7:27, 8:20},
    3: {1:15, 2:38, 4:20, 5:21, 6:21, 7:16, 8:25},
    4: {1:21, 2:30, 3:20, 5:33, 6:15, 7:23, 8:20},
    5: {1:27, 2:37, 3:21, 4:33, 6:25, 7:13, 8:25},
    6: {1:18, 2:23, 3:21, 4:15, 5:25, 7:26, 8:14},
    7: {1:19, 2:27, 3:16, 4:23, 5:13, 6:26, 8:30},
    8: {1:23, 2:20, 3:25, 4:20, 5:25, 6:14, 7:30}
}

# 3. 무방향 그래프 생성 및 간선 추가
G = nx.Graph()
for u in matrix:
    for v in matrix[u]:
        if u < v:  # 중복 간선 방지
            G.add_edge(u, v, weight=matrix[u][v])

# 4. 그래프 시각화 레이아웃 설정
plt.figure(figsize=(11, 9))
pos = nx.circular_layout(G)

# 노드 그리기 (글자가 안 가려지도록 크기를 3200으로 확대)
nx.draw_networkx_nodes(G, pos, node_color='#8DA0CB', edgecolors='black', node_size=3200)


nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_weight='bold', font_family=font_name)

# 간선 그리기
nx.draw_networkx_edges(G, pos, edge_color='silver', width=1.5)

# 간선 가중치(이동 시간) 표시 (숫자만 표시, 폰트 적용)
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color='red', font_size=10, font_weight='bold', font_family=font_name)

plt.title("모든 음식점 이동 시간 그래프", fontsize=16, fontweight='bold', pad=20)
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import platform

# 1. 한글 폰트 깨짐 방지
system_name = platform.system()
if system_name == 'Windows':
    plt.rc('font', family='Malgun Gothic')
elif system_name == 'Darwin':
    plt.rc('font', family='AppleGothic')
else:
    plt.rc('font', family='NanumGothic')
plt.rcParams['axes.unicode_minus'] = False

# 2. 노드 및 간선 설정
nodes = [1, 2, 3, 4, 5, 6, 7, 8]

labels = {
    1: "멜트", 2: "방목", 3: "용산토리", 4: "동녘",
    5: "초이노미", 6: "이화만두", 7: "상상홍익", 8: "팔도실비"
}

edges = [
    (1, 2), (1, 6), (1, 8), (2, 3), (2, 5), (2, 6),
    (3, 4), (3, 5), (3, 6), (4, 5), (6, 7), (6, 8)
]

pos = {
    1: (-1, 2), 2: (-2, 0), 3: (-1, -2), 4: (2, -2),
    5: (1, 0), 6: (0, 1), 7: (2, 1.5), 8: (0, 3)
}

G = nx.Graph()
G.add_nodes_from(nodes)
G.add_edges_from(edges)

# 3. DFS 진행 순서 하드코딩 (1->2->3->4->5 -> 6->7 -> 8)
dfs_steps = [
    {'visited': [1], 'edges': []},
    {'visited': [1, 2], 'edges': [(1, 2)]},
    {'visited': [1, 2, 3], 'edges': [(1, 2), (2, 3)]},
    {'visited': [1, 2, 3, 4], 'edges': [(1, 2), (2, 3), (3, 4)]},
    {'visited': [1, 2, 3, 4, 5], 'edges': [(1, 2), (2, 3), (3, 4), (4, 5)]},
    {'visited': [1, 2, 3, 4, 5, 6], 'edges': [(1, 2), (2, 3), (3, 4), (4, 5), (3, 6)]},
    {'visited': [1, 2, 3, 4, 5, 6, 7], 'edges': [(1, 2), (2, 3), (3, 4), (4, 5), (3, 6), (6, 7)]},
    {'visited': [1, 2, 3, 4, 5, 6, 7, 8], 'edges': [(1, 2), (2, 3), (3, 4), (4, 5), (3, 6), (6, 7), (6, 8)]}
]

fig, ax = plt.subplots(figsize=(10, 8))

def update_dfs(frame):
    ax.clear()
    ax.axis('off')
    data = dfs_steps[frame]

    # 방문한 노드는 파란색
    node_colors = ['#8DA0CB' if n in data['visited'] else 'white' for n in nodes]
    nx.draw_networkx_nodes(G, pos, node_color=node_colors, edgecolors='black', node_size=1500, ax=ax)
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=11, font_weight='bold', font_family=plt.rcParams['font.family'], ax=ax)

    # 지나간 간선은 굵은 검은색
    edge_colors = ['black' if (u, v) in data['edges'] or (v, u) in data['edges'] else 'lightgray' for u, v in G.edges()]
    edge_widths = [2.5 if (u, v) in data['edges'] or (v, u) in data['edges'] else 1.0 for u, v in G.edges()]
    nx.draw_networkx_edges(G, pos, edge_color=edge_colors, width=edge_widths, ax=ax)

    ax.text(0.5, 1.05, f"DFS 탐색 단계: {frame + 1} / {len(dfs_steps)}", transform=ax.transAxes, ha='center', fontsize=16, fontweight='bold')

ani_dfs = animation.FuncAnimation(fig, update_dfs, frames=len(dfs_steps), interval=1500, repeat=False)
plt.tight_layout()
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import platform

# 1. 한글 폰트 깨짐 방지
system_name = platform.system()
if system_name == 'Windows':
    plt.rc('font', family='Malgun Gothic')
elif system_name == 'Darwin':
    plt.rc('font', family='AppleGothic')
else:
    plt.rc('font', family='NanumGothic')
plt.rcParams['axes.unicode_minus'] = False

# 2. 노드 및 간선 설정
nodes = [1, 2, 3, 4, 5, 6, 7, 8]


labels = {
    1: "멜트", 2: "방목", 3: "용산토리", 4: "동녘",
    5: "초이노미", 6: "이화만두", 7: "상상홍익", 8: "팔도실비"
}

edges = [
    (1, 2), (1, 6), (1, 8), (2, 3), (2, 5), (2, 6),
    (3, 4), (3, 5), (3, 6), (4, 5), (6, 7), (6, 8)
]

pos = {
    1: (-1, 2), 2: (-2, 0), 3: (-1, -2), 4: (2, -2),
    5: (1, 0), 6: (0, 1), 7: (2, 1.5), 8: (0, 3)
}

G = nx.Graph()
G.add_nodes_from(nodes)
G.add_edges_from(edges)


bfs_steps = [
    # 1단계: 시작 노드 멜트(1) 방문
    {'step': '멜트 방문', 'visited': [1], 'active_nodes': [1], 'edges': [], 'active_edges': []},

    # 2단계: 멜트(1)에서 팔도실비(8) 방문
    {'step': '팔도실비 방문 (from 멜트)', 'visited': [1, 8], 'active_nodes': [8],
     'edges': [(1, 8)], 'active_edges': [(1, 8)]},

    # 3단계: 멜트(1)에서 이화만두(6) 방문
    {'step': '이화만두 방문 (from 멜트)', 'visited': [1, 8, 6], 'active_nodes': [6],
     'edges': [(1, 8), (1, 6)], 'active_edges': [(1, 6)]},

    # 4단계: 멜트(1)에서 방목(2) 방문
    {'step': '방목 방문 (from 멜트)', 'visited': [1, 8, 6, 2], 'active_nodes': [2],
     'edges': [(1, 8), (1, 6), (1, 2)], 'active_edges': [(1, 2)]},

    # 5단계: 이화만두(6)에서 상상홍익(7) 방문
    {'step': '상상홍익 방문 (from 이화만두)', 'visited': [1, 8, 6, 2, 7], 'active_nodes': [7],
     'edges': [(1, 8), (1, 6), (1, 2), (6, 7)], 'active_edges': [(6, 7)]},

    # 6단계: 방목(2)에서 초이노미(5) 방문
    {'step': '초이노미 방문 (from 방목)', 'visited': [1, 8, 6, 2, 7, 5], 'active_nodes': [5],
     'edges': [(1, 8), (1, 6), (1, 2), (6, 7), (2, 5)], 'active_edges': [(2, 5)]},

    # 7단계: 방목(2)에서 용산토리(3) 방문
    {'step': '용산토리 방문 (from 방목)', 'visited': [1, 8, 6, 2, 7, 5, 3], 'active_nodes': [3],
     'edges': [(1, 8), (1, 6), (1, 2), (6, 7), (2, 5), (2, 3)], 'active_edges': [(2, 3)]},

    # 8단계: 초이노미(5)에서 동녘(4) 방문 (초이노미 간선 연결)
    {'step': '동녘 방문 (from 초이노미)', 'visited': [1, 8, 6, 2, 7, 5, 3, 4], 'active_nodes': [4],
     'edges': [(1, 8), (1, 6), (1, 2), (6, 7), (2, 5), (2, 3), (5, 4)], 'active_edges': [(5, 4)]},

    # 9단계: 탐색 종료 및 최종 트리 완성 상태
    {'step': '모든 노드 탐색 완료', 'visited': [1, 8, 6, 2, 7, 5, 3, 4], 'active_nodes': [],
     'edges': [(1, 8), (1, 6), (1, 2), (6, 7), (2, 5), (2, 3), (5, 4)],
     'active_edges': [(1, 8), (1, 6), (1, 2), (6, 7), (2, 5), (2, 3), (5, 4)]}
]

fig, ax = plt.subplots(figsize=(10, 8))

def update_bfs(frame):
    ax.clear()
    ax.axis('off')
    data = bfs_steps[frame]

    node_colors = []
    node_edge_colors = []
    node_linewidths = []

    # 노드 스타일 설정
    for n in nodes:
        if n in data['active_nodes']:
            node_colors.append('#8DA0CB')  # 현재 활성화된 방문 노드 (파란색)
            node_edge_colors.append('black')
            node_linewidths.append(1.5)
        elif n in data['visited']:
            node_colors.append('white')    # 이미 방문이 끝난 노드 (흰색)
            node_edge_colors.append('black')
            node_linewidths.append(3.0)    # 굵은 테두리로 표시
        else:
            node_colors.append('white')    # 아직 탐색 전인 노드
            node_edge_colors.append('gray')
            node_linewidths.append(1.0)

    nx.draw_networkx_nodes(G, pos, node_color=node_colors, edgecolors=node_edge_colors, linewidths=node_linewidths, node_size=1600, ax=ax)
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=11, font_weight='bold', font_family=plt.rcParams['font.family'], ax=ax)

    edge_colors = []
    edge_widths = []

    # 간선 스타일 설정
    for u, v in G.edges():
        if (u, v) in data['active_edges'] or (v, u) in data['active_edges']:
            edge_colors.append('#8DA0CB')  # 현재 탐색/이동 중인 간선 (파란색)
            edge_widths.append(4.0)
        elif (u, v) in data['edges'] or (v, u) in data['edges']:
            edge_colors.append('black')    # 확정된 BFS 트리 경로 (검은색)
            edge_widths.append(4.0)
        else:
            edge_colors.append('lightgray')
            edge_widths.append(1.5)

    nx.draw_networkx_edges(G, pos, edge_color=edge_colors, width=edge_widths, ax=ax)

    ax.text(0.5, 1.05, f"BFS 순차 탐색: {data['step']}", transform=ax.transAxes, ha='center', fontsize=16, fontweight='bold')

ani_bfs = animation.FuncAnimation(fig, update_bfs, frames=len(bfs_steps), interval=800, repeat=False)
plt.tight_layout()
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import platform

# 1. 한글 폰트 깨짐 방지 설정
system_name = platform.system()
if system_name == 'Windows':
    plt.rc('font', family='Malgun Gothic')
elif system_name == 'Darwin':
    plt.rc('font', family='AppleGothic')
else:
    plt.rc('font', family='NanumGothic')
plt.rcParams['axes.unicode_minus'] = False

# 2. 노드 및 음식점 라벨 설정 (숫자 제외)
nodes = [1, 2, 3, 4, 5, 6, 7, 8]
labels = {
    1: "멜트", 2: "방목", 3: "용산토리", 4: "동녘",
    5: "초이노미", 6: "이화만두", 7: "상상홍익", 8: "팔도실비"
}

matrix = {
    1: {2:34, 3:15, 4:21, 5:27, 6:18, 7:19, 8:23},
    2: {1:34, 3:38, 4:30, 5:37, 6:23, 7:27, 8:20},
    3: {1:15, 2:38, 4:20, 5:21, 6:21, 7:16, 8:25},
    4: {1:21, 2:30, 3:20, 5:33, 6:15, 7:23, 8:20},
    5: {1:27, 2:37, 3:21, 4:33, 6:25, 7:13, 8:25},
    6: {1:18, 2:23, 3:21, 4:15, 5:25, 7:26, 8:14},
    7: {1:19, 2:27, 3:16, 4:23, 5:13, 6:26, 8:30},
    8: {1:23, 2:20, 3:25, 4:20, 5:25, 6:14, 7:30}
}

# 그래프 생성 및 28개 간선(Complete Graph) 전체 추가
G = nx.Graph()
G.add_nodes_from(nodes)
for u in matrix:
    for v in matrix[u]:
        if u < v:
            G.add_edge(u, v, weight=matrix[u][v])

# 가독성을 극대화하기 위해 원형(Circular) 배치 사용
pos = nx.circular_layout(G)

# 3. 프림 알고리즘 진행 단계 하드코딩 (시작점: 멜트)
prim_steps = [
    {'step': '0', 'title': '시작 정점 선택: [멜트]', 'S': [1], 'mst_edges': [], 'active_node': 1, 'active_edge': None},
    {'step': '1', 'title': '[용산토리] 방문 (연결: 멜트, 가중치: 15)', 'S': [1, 3], 'mst_edges': [(1, 3)], 'active_node': 3, 'active_edge': (1, 3)},
    {'step': '2', 'title': '[상상홍익] 방문 (연결: 용산토리, 가중치: 16)', 'S': [1, 3, 7], 'mst_edges': [(1, 3), (3, 7)], 'active_node': 7, 'active_edge': (3, 7)},
    {'step': '3', 'title': '[초이노미] 방문 (연결: 상상홍익, 가중치: 13)', 'S': [1, 3, 7, 5], 'mst_edges': [(1, 3), (3, 7), (7, 5)], 'active_node': 5, 'active_edge': (7, 5)},
    {'step': '4', 'title': '[이화만두] 방문 (연결: 멜트, 가중치: 18)', 'S': [1, 3, 7, 5, 6], 'mst_edges': [(1, 3), (3, 7), (7, 5), (1, 6)], 'active_node': 6, 'active_edge': (1, 6)},
    {'step': '5', 'title': '[팔도실비] 방문 (연결: 이화만두, 가중치: 14)', 'S': [1, 3, 7, 5, 6, 8], 'mst_edges': [(1, 3), (3, 7), (7, 5), (1, 6), (6, 8)], 'active_node': 8, 'active_edge': (6, 8)},
    {'step': '6', 'title': '[동녘] 방문 (연결: 이화만두, 가중치: 15)', 'S': [1, 3, 7, 5, 6, 8, 4], 'mst_edges': [(1, 3), (3, 7), (7, 5), (1, 6), (6, 8), (6, 4)], 'active_node': 4, 'active_edge': (6, 4)},
    {'step': '7', 'title': '[방목] 방문 (연결: 팔도실비, 가중치: 20)', 'S': [1, 3, 7, 5, 6, 8, 4, 2], 'mst_edges': [(1, 3), (3, 7), (7, 5), (1, 6), (6, 8), (6, 4), (8, 2)], 'active_node': 2, 'active_edge': (8, 2)},
    {'step': '8', 'title': '최소 신장 트리(MST) 완성 완료', 'S': [1, 3, 7, 5, 6, 8, 4, 2], 'mst_edges': [(1, 3), (3, 7), (7, 5), (1, 6), (6, 8), (6, 4), (8, 2)], 'active_node': None, 'active_edge': None}
]

fig, ax = plt.subplots(figsize=(11, 9))

def update_prim(frame):
    ax.clear()
    ax.axis('off')
    data = prim_steps[frame]

    # --- 1. 노드 스타일링 ---
    node_colors = []
    node_edges = []
    node_widths = []

    for n in nodes:
        if n == data['active_node']:
            node_colors.append('#FF9999')  # 방금 추가된 정점 (연분홍)
            node_edges.append('red')
            node_widths.append(2.5)
        elif n in data['S']:
            node_colors.append('#8DA0CB')  # MST 집합 S에 포함된 정점 (파란색)
            node_edges.append('black')
            node_widths.append(2.0)
        else:
            node_colors.append('white')    # 아직 선택되지 않은 정점
            node_edges.append('gray')
            node_widths.append(1.0)

    nx.draw_networkx_nodes(G, pos, node_color=node_colors, edgecolors=node_edges, linewidths=node_widths, node_size=1800, ax=ax)
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=11, font_weight='bold', font_family=plt.rcParams['font.family'], ax=ax)

    # --- 2. 간선 스타일링 및 가중치 표시 ---
    non_mst_edges = []
    mst_edges = []
    active_edge = []

    for u, v in G.edges():
        if data['active_edge'] and ((u, v) == data['active_edge'] or (v, u) == data['active_edge']):
            active_edge.append((u, v))
        elif (u, v) in data['mst_edges'] or (v, u) in data['mst_edges']:
            mst_edges.append((u, v))
        else:
            non_mst_edges.append((u, v))

    # 일반 간선 (연하게 처리하여 가독성 확보)
    nx.draw_networkx_edges(G, pos, edgelist=non_mst_edges, edge_color='gainsboro', width=1.0, alpha=0.4, ax=ax)
    # 확정된 MST 간선
    nx.draw_networkx_edges(G, pos, edgelist=mst_edges, edge_color='black', width=3.5, ax=ax)
    # 현재 선택된 간선 (적색 강조)
    if active_edge:
        nx.draw_networkx_edges(G, pos, edgelist=active_edge, edge_color='red', width=4.5, ax=ax)

    # --- 3. 가중치 텍스트 스타일 분리 ---
    all_weights = nx.get_edge_attributes(G, 'weight')

    non_mst_labels = {k: v for k, v in all_weights.items() if k in non_mst_edges or (k[1], k[0]) in non_mst_edges}
    mst_labels = {k: v for k, v in all_weights.items() if k in mst_edges or (k[1], k[0]) in mst_edges}
    active_labels = {k: v for k, v in all_weights.items() if k in active_edge or (k[1], k[0]) in active_edge}

    # 일반 가중치는 작고 투명하게 기본 배치
    nx.draw_networkx_edge_labels(G, pos, edge_labels=non_mst_labels, font_size=8, font_color='gray', alpha=0.5, ax=ax)
    # MST 경로 가중치는 크고 선명하게 배치
    nx.draw_networkx_edge_labels(G, pos, edge_labels=mst_labels, font_size=11, font_color='black', font_weight='bold', ax=ax)
    if active_labels:
        nx.draw_networkx_edge_labels(G, pos, edge_labels=active_labels, font_size=12, font_color='red', font_weight='bold', ax=ax)

    # 상단 타이틀 표시
    ax.text(0.5, 1.05, f"프림(Prim) 알고리즘 구현: {data['title']}", transform=ax.transAxes, ha='center', fontsize=16, fontweight='bold')

# 애니메이션 초기화 및 구동 (간격: 1.2초)
ani_prim = animation.FuncAnimation(fig, update_prim, frames=len(prim_steps), interval=1200, repeat=False)
plt.tight_layout()

# 화면에 애니메이션 창 띄우기
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import platform

# 1. 한글 폰트 깨짐 방지 설정
system_name = platform.system()
if system_name == 'Windows':
    plt.rc('font', family='Malgun Gothic')
elif system_name == 'Darwin':
    plt.rc('font', family='AppleGothic')
else:
    plt.rc('font', family='NanumGothic')
plt.rcParams['axes.unicode_minus'] = False

# 2. 노드 및 음식점 라벨 설정 (숫자 제외)
nodes = [1, 2, 3, 4, 5, 6, 7, 8]
labels = {
    1: "멜트", 2: "방목", 3: "용산토리", 4: "동녘",
    5: "초이노미", 6: "이화만두", 7: "상상홍익", 8: "팔도실비"
}

# 사진 데이터 기반 인접 행렬 (이동 시간 가중치)
matrix = {
    1: {2:34, 3:15, 4:21, 5:27, 6:18, 7:19, 8:23},
    2: {1:34, 3:38, 4:30, 5:37, 6:23, 7:27, 8:20},
    3: {1:15, 2:38, 4:20, 5:21, 6:21, 7:16, 8:25},
    4: {1:21, 2:30, 3:20, 5:33, 6:15, 7:23, 8:20},
    5: {1:27, 2:37, 3:21, 4:33, 6:25, 7:13, 8:25},
    6: {1:18, 2:23, 3:21, 4:15, 5:25, 7:26, 8:14},
    7: {1:19, 2:27, 3:16, 4:23, 5:13, 6:26, 8:30},
    8: {1:23, 2:20, 3:25, 4:20, 5:25, 6:14, 7:30}
}

# 그래프 생성 및 28개 간선(Complete Graph) 전체 추가
G = nx.Graph()
G.add_nodes_from(nodes)
for u in matrix:
    for v in matrix[u]:
        if u < v:
            G.add_edge(u, v, weight=matrix[u][v])

pos = nx.circular_layout(G)

# 3. 크루스칼 알고리즘 진행 단계 하드코딩 (가중치 오름차순 검사)
kruskal_steps = [
    {'title': '크루스칼 알고리즘 시작 (가중치 오름차순 검사)', 'mst_edges': [], 'testing_edge': None, 'status': 'start'},

    {'title': '가중치 13 (초이노미-상상홍익) 검사 -> 채택', 'mst_edges': [(5,7)], 'testing_edge': (5,7), 'status': 'accept'},

    {'title': '가중치 14 (이화만두-팔도실비) 검사 -> 채택', 'mst_edges': [(5,7), (6,8)], 'testing_edge': (6,8), 'status': 'accept'},

    {'title': '가중치 15 (멜트-용산토리) 검사 -> 채택', 'mst_edges': [(5,7), (6,8), (1,3)], 'testing_edge': (1,3), 'status': 'accept'},

    {'title': '가중치 15 (동녘-이화만두) 검사 -> 채택', 'mst_edges': [(5,7), (6,8), (1,3), (4,6)], 'testing_edge': (4,6), 'status': 'accept'},

    {'title': '가중치 16 (용산토리-상상홍익) 검사 -> 채택', 'mst_edges': [(5,7), (6,8), (1,3), (4,6), (3,7)], 'testing_edge': (3,7), 'status': 'accept'},

    {'title': '가중치 18 (멜트-이화만두) 검사 -> 채택', 'mst_edges': [(5,7), (6,8), (1,3), (4,6), (3,7), (1,6)], 'testing_edge': (1,6), 'status': 'accept'},

    # 사이클이 발생하는 간선들 (거절됨)
    {'title': '가중치 19 (멜트-상상홍익) 검사 -> 거절 (사이클 발생)', 'mst_edges': [(5,7), (6,8), (1,3), (4,6), (3,7), (1,6)], 'testing_edge': (1,7), 'status': 'reject'},
    {'title': '가중치 20 (용산토리-동녘) 검사 -> 거절 (사이클 발생)', 'mst_edges': [(5,7), (6,8), (1,3), (4,6), (3,7), (1,6)], 'testing_edge': (3,4), 'status': 'reject'},
    {'title': '가중치 20 (동녘-팔도실비) 검사 -> 거절 (사이클 발생)', 'mst_edges': [(5,7), (6,8), (1,3), (4,6), (3,7), (1,6)], 'testing_edge': (4,8), 'status': 'reject'},

    # 마지막 간선 채택으로 7개 간선 완성
    {'title': '가중치 20 (방목-팔도실비) 검사 -> 채택', 'mst_edges': [(5,7), (6,8), (1,3), (4,6), (3,7), (1,6), (2,8)], 'testing_edge': (2,8), 'status': 'accept'},

    {'title': '최소 신장 트리(MST) 완성 완료', 'mst_edges': [(5,7), (6,8), (1,3), (4,6), (3,7), (1,6), (2,8)], 'testing_edge': None, 'status': 'done'}
]

fig, ax = plt.subplots(figsize=(11, 9))
rejected_edges = set()

def update_kruskal(frame):
    ax.clear()
    ax.axis('off')
    data = kruskal_steps[frame]

    # 거절된 간선을 기록해두어 화면에 연한 빨간색 점선으로 남김
    if data['status'] == 'reject' and data['testing_edge']:
        rejected_edges.add(data['testing_edge'])

    # --- 1. 노드 스타일링 ---
    nx.draw_networkx_nodes(G, pos, node_color='white', edgecolors='black', linewidths=1.5, node_size=1800, ax=ax)
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=11, font_weight='bold', font_family=plt.rcParams['font.family'], ax=ax)

    # --- 2. 간선 분류 및 스타일링 ---
    non_mst_edges = []
    mst_edges = []

    for u, v in G.edges():
        if (u, v) in data['mst_edges'] or (v, u) in data['mst_edges']:
            mst_edges.append((u, v))
        elif (u, v) not in rejected_edges and (v, u) not in rejected_edges:
            # 테스트 중인 간선 제외하고 기본 회색선
            if data['testing_edge'] and ((u, v) == data['testing_edge'] or (v, u) == data['testing_edge']):
                continue
            non_mst_edges.append((u, v))

    # 일반 대기 간선 (연한 회색)
    nx.draw_networkx_edges(G, pos, edgelist=non_mst_edges, edge_color='gainsboro', width=1.0, alpha=0.4, ax=ax)
    # 이미 거절된 간선 (연한 붉은 점선)
    if rejected_edges:
        nx.draw_networkx_edges(G, pos, edgelist=list(rejected_edges), edge_color='salmon', width=1.5, style='dashed', alpha=0.5, ax=ax)
    # 확정된 MST 간선 (검정색)
    nx.draw_networkx_edges(G, pos, edgelist=mst_edges, edge_color='black', width=3.5, ax=ax)

    # 현재 테스트 중인 간선 (채택: 파란색, 거절: 빨간색 강조)
    if data['testing_edge']:
        edge_color = 'royalblue' if data['status'] == 'accept' else 'red'
        nx.draw_networkx_edges(G, pos, edgelist=[data['testing_edge']], edge_color=edge_color, width=4.5, ax=ax)

    # --- 3. 가중치 텍스트 표시 (숫자만 표시) ---
    all_weights = nx.get_edge_attributes(G, 'weight')

    non_mst_labels = {k: v for k, v in all_weights.items() if k in non_mst_edges or (k[1], k[0]) in non_mst_edges}
    mst_labels = {k: v for k, v in all_weights.items() if k in mst_edges or (k[1], k[0]) in mst_edges}
    testing_labels = {k: v for k, v in all_weights.items() if data['testing_edge'] and (k == data['testing_edge'] or (k[1], k[0]) == data['testing_edge'])}

    nx.draw_networkx_edge_labels(G, pos, edge_labels=non_mst_labels, font_size=8, font_color='gray', alpha=0.5, ax=ax)
    nx.draw_networkx_edge_labels(G, pos, edge_labels=mst_labels, font_size=11, font_color='black', font_weight='bold', ax=ax)

    if testing_labels:
        font_color = 'royalblue' if data['status'] == 'accept' else 'red'
        nx.draw_networkx_edge_labels(G, pos, edge_labels=testing_labels, font_size=13, font_color=font_color, font_weight='bold', ax=ax)

    # 상단 타이틀
    title_color = 'black'
    if data['status'] == 'accept': title_color = 'royalblue'
    elif data['status'] == 'reject': title_color = 'red'

    ax.text(0.5, 1.05, f"크루스칼(Kruskal) 진행: {data['title']}", transform=ax.transAxes, ha='center', fontsize=15, fontweight='bold', color=title_color)

# 애니메이션 셋팅 (간격 1.8초로 여유있게 지정)
ani_kruskal = animation.FuncAnimation(fig, update_kruskal, frames=len(kruskal_steps), interval=1800, repeat=False)
plt.tight_layout()
plt.show()
```
